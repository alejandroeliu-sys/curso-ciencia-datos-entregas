```python
import time, duckdb, pandas as pd

t = time.time()
csv = duckdb.sql("""SELECT year(START) AS anio, count(*) n
  FROM read_csv_auto('/home/alecedillo/synthea/output/csv/encounters.csv') GROUP BY 1 ORDER BY 1 DESC LIMIT 2""").df() 
t_csv = time.time() - t

t = time.time()
pq = duckdb.sql("""SELECT anio, count(*) n
  FROM read_parquet('parquet_local/encounters/*/*.parquet', hive_partitioning=true)
  GROUP BY 1 ORDER BY 1 DESC LIMIT 2""").df()
t_pq = time.time() - t

print(csv, pq, sep="\n")
pd.DataFrame({"formato": ["CSV", "Parquet"], "segundos": [t_csv, t_pq]})
```

```python
print(f"Parquet fue {t_csv/t_pq:.1f}x más rápido que CSV (CSV 1,309 MB vs Parquet 155 MB)")
```
