```python
import boto3

session = boto3.Session(profile_name="tesis")   # pista: el nombre del perfil
s3 = session.client("s3")
print(session.region_name)
```

```python
respuesta = s3.list_buckets()
for b in respuesta["Buckets"]:
    print(b["Name"])   # pista: "Name"
```

```python
bucket = "alejandro-lab19-20261020"
clave = "publico/archivo.txt"   # pista: el archivo que subiste

s3.download_file(bucket, clave, "descargado.txt")
print(open("descargado.txt").read())
```
